# ¿Qué contiene este dataset?

Este conjunto de datos contiene 1.000.000 de filas de datos sobre el rendimiento estudiantil.

Cada fila representa a un estudiante con características como horas de estudio, asistencia, participación en clase y calificación final.

| Variable (Columna) | Descripción |
|---------|-------------|
| **student_id** | Identificador único para cada estudiante. |
| **weekly_self_study_hours (horas_de_autoaprendizaje_semanales)** | Promedio de horas semanales de autoaprendizaje (0–40).  |
| **attendance_percentage  (porcentaje_de_asistencia)** | Porcentaje de asistencia (50–100).  |
| **class_participation (participación_en_clase)** | Puntuación entre 0 y 10 que indica el grado de participación del estudiante en clase.|
| **total_score (puntuación_total)** | Puntuación de rendimiento final (0–100).  |
| **grade (calificación)** | Etiqueta categórica (A, B, C, D, F) derivada de la puntuación total. |
